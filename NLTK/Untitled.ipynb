```python
encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(removed_stopwords)

print(encoder.get_feature_names_out())
print(encoded)
```

```python
document = " ".join(w for w in removed_stopwords " ".join(removed_stopwords))

corpus = [document]
```

```python
cv = CountVectorizer()

X = cv.fit_transform(corpus)

print("Vocabulary:", cv.get_feature_names_out())
print("Vectors:\n", X.toarray())
```

```python
ngram = CountVectorizer(ngram_range=(1, 2))
X_ngram = ngram.fit_transform(corpus)

print("Features:", ngram.get_feature_names_out())
print(X_ngram.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

X_tfidf = tfidf.fit_transform(corpus)

print("Features:", tfidf.get_feature_names_out())
print("TF-IDF Matrix:\n", X_tfidf.toarray())
```
